# 17 · Cross-dataset synthesis

Panels that draw on *several* datasets at once, so they don't belong to any single dataset notebook. Each cell regenerates a standalone script under `../manuscript/figures/fig_extra/`.

- **Transcriptomic cross-check** — independent convergent validity: HOMER's π (connectivity + anchors, never genes) routes to transcriptomically similar human cortex at z≈10 vs a 51-gene homolog panel.
- **Conservation spectrum** — every result placed by evidence strength × granularity; discrete & contrastive signals translate, diffuse continuous ones wash out.
- **Connectional, not molecular** — the homology boundary is a coverage (circuit) phenomenon: transcriptomic similarity does *not* collapse over association cortex.
- **Beauchamp systems ladder + Whitesell DMN** — which homologies validate, and how refining the mouse-DMN definition sharpens the DMN→DMN correspondence.
